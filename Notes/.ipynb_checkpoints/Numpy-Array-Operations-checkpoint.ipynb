# Array Operations
## Arthmetic Operation

```python
a1 = np.array([1, 2, 3, 4, 5])
a2 = np.array([6, 7, 8, 9, 0])
```

```python
arr1 = a1 + a2
arr1
```

```python
arr2 = a1 - a2
arr2
```

```python
arr3 = a1 * a2
arr3
```

```python
arr4 = a2 / a1
arr4
```

```python
arr5 = a2 // a1
arr5
```

```python
arr6 = a1 ** a2
arr6
```

## Broadcasting

```python
l = [10, 20, 30, 40]
arr = np.array(l)
```

```python
for i in l:
    print(i + 10)
```

```python
add = arr + 10
add
```

```python
arr2 = np.arange(1,26).reshape(5,5)
arr2
```

```python
arr2 + 10
```

## Deep & Shallow copy

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice
```

```python
slice = slice * 10
slice
```

```python
a
```

```python
a[4:6] * 10
```

```python
a
```

```python
b = a.copy()
```

```python
b[0] = 90
b
```

```python
a
```

```python
A
```

```python
B
```

# Advance Arry Manipulation
## Stacking Array

```python
a1 = np.array([1, 2, 3, 4, 5])
a2 = np.array([6, 7, 8, 9, 0])
```

```python
np.vstack((a1,a2))
```

```python
np.hstack((a1,a2))
```

```python
np.column_stack((a1,a2))
```

## Splitting Array

```python
c = np.arange(16).reshape(4, 4)
c
```

```python
np.hsplit(c, 2)
```

```python
a = np.vsplit(c, 2)
a
```

```python
for i in a:
    print(i)
```
