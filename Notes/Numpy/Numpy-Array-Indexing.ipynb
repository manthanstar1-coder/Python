```python
import numpy as np
```

## Numpy Indexing and Slicing of Vectors

```python
arr = np.arange(11,21)
arr
```

```python
arr[7]
```

```python
arr_slice = arr[4::2]
arr_slice
```

## Numpy Indexing and Slicing of Matrix

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[5,4]
```

```python
slice = arr[0:2,1:3]
slice
```

```python
slice1 = arr[3:,3:]
slice1
```

```python
slice2 = arr[:,2]
slice2
```

## Boolean Indexing

```python
arr = np.arange(11,31)
arr
```

```python
bool_ind = arr % 2 == 0
bool_ind
```

```python
arr = arr[bool_ind]
arr
```
